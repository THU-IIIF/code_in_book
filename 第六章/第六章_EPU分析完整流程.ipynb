# 第六章：EPU 指数生成与分析（Notebook 版）

本 Notebook 覆盖以下流程：

1. 运行 `1_EPU_generation.py` 生成 EPU 指数及市场数据。
2. 运行 `2_EPU_analysis.py` 生成表格、图表和分析摘要。
3. 检查输出文件并在 Notebook 中预览关键结果。

> 说明：Notebook 会直接复用脚本中的函数，保证与命令行输出保持一致。


```python
import os
import importlib.util
from pathlib import Path

import pandas as pd
from IPython.display import display, Image

# 确认工作目录指向 chapter6_code
BASE_DIR = Path.cwd().resolve()
if not (BASE_DIR / "1_EPU_generation.py").exists():
    candidate = BASE_DIR / "chapter6_code"
    if (candidate / "1_EPU_generation.py").exists():
        os.chdir(candidate)
        BASE_DIR = candidate.resolve()

OUTPUT_DIR = BASE_DIR / "output"
OUTPUT_DIR.mkdir(exist_ok=True)
print(f"当前工作目录: {BASE_DIR}")
```

```python
# 配置 Matplotlib 与 Fontconfig 缓存目录
mpl_dir = OUTPUT_DIR / ".matplotlib"
cache_dir = OUTPUT_DIR / ".cache"
font_dir = cache_dir / "fontconfig"

for path in [mpl_dir, cache_dir, font_dir]:
    path.mkdir(parents=True, exist_ok=True)

os.environ.setdefault("MPLBACKEND", "Agg")
os.environ.setdefault("MPLCONFIGDIR", str(mpl_dir))
os.environ.setdefault("XDG_CACHE_HOME", str(cache_dir))
os.environ.setdefault("FC_CACHEDIR", str(font_dir))

print("Matplotlib 缓存目录:", mpl_dir)
print("Fontconfig 缓存目录:", font_dir)
```

```python
def load_module(name: str, file_path: Path):
    """动态加载指定 Python 脚本为模块。"""
    spec = importlib.util.spec_from_file_location(name, str(file_path))
    module = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(module)
    return module

print("辅助函数已加载。")
```

```python
# 运行程序 1：生成 EPU 指数及市场数据
epu_generation = load_module("epu_generation", BASE_DIR / "1_EPU_generation.py")
epu_generation.main()

epu_data_path = BASE_DIR / "data" / "epu_index.csv"
assert epu_data_path.exists(), "数据文件未生成"
print(f"数据文件位置: {epu_data_path}")
```

```python
epu_df = pd.read_csv(epu_data_path)
display(epu_df.head())
print(f"数据维度: {epu_df.shape}")
```

```python
# 运行程序 2：生成表格、图表与分析摘要
epu_analysis = load_module("epu_analysis", BASE_DIR / "2_EPU_analysis.py")
epu_analysis.main()

summary_path = OUTPUT_DIR / "cell07_分析摘要" / "analysis_summary.txt"
assert summary_path.exists(), "分析摘要未生成"
print(f"分析摘要位置: {summary_path}")
```

```python
with summary_path.open("r", encoding="utf-8") as f:
    print(f.read())
```
